```python
def perceptron(x1,x2,w1,w2,b):
    z = w1*x1+w2*x2+b

    if z>=0:
        y_pred = 1
    else:
        y_pred = 0
    return y_pred
    
```

```python
def sigmoid(self, x):
    return 1 / (1 + np.exp(-x))
    
```

```python
def sigmoid_derivative(self, x):
    return x *(1-x)
```

```python
def forward(self,X):
    self.z1 = np.dot(X.self.W1)+self.b1
    self.a1 = self.sigmoid(self.z1)
    self.z2 = np.dot(self.a1, self.W2) + self.b2
    self.a2 = self.sigmoid(self.z2)
    return self.a2
```
